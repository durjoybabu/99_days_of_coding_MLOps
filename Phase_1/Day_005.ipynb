```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv('/content/NIFTY-26SEP24-09JUL24.csv')
print("Data loaded successfully!")
```

```python

print("--- First 5 rows (head) ---")
display(data.head())

print("\n--- Last 5 rows (tail) ---")
display(data.tail())

print(f"\n--- Shape of the data ---\n{data.shape}")

print("\n--- Information about the dataset ---")
data.info()
```

```python

print("--- Summary Statistics ---")
display(data.describe())

print("\n--- Min Values (Numeric Only) ---")
print(data.min(numeric_only=True))

print("\n--- Max Values (Numeric Only) ---")
print(data.max(numeric_only=True))

print("\n--- Unique values in 'right' column ---")
print(data['right'].unique())

print("\n--- Value counts for 'right' column ---")
print(data['right'].value_counts())
```

```python

print("--- Missing values count in each column ---")
print(data.isnull().sum())

print("\n--- Duplicate entries count ---")
duplicate_count = data.duplicated().sum()
print(f"Total duplicates: {duplicate_count}")

if duplicate_count > 0:
    print("Dropping duplicates...")
    data = data.drop_duplicates()
    print(f"New shape: {data.shape}")
```

```python

print("--- Plotting Histograms for Numeric Columns ---")
data.hist(figsize=(12, 10), bins=30)
plt.tight_layout()
plt.show()

print("\n--- Boxplot of Closing Price by Options Right (CE/PE) ---")
plt.figure(figsize=(8, 5))
sns.boxplot(x='right', y='close', data=data)
plt.title('Closing Price Distribution by CE/PE')
plt.show()
```
