```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


df_new = pd.read_csv('/content/NIFTY-26SEP24-12JUL24.csv')


display(df_new.head())
display(df_new.info())
```

```python

numerical_cols = ['open', 'high', 'low', 'close', 'open_interest', 'volume', 'strike_price']
corr_matrix = df_new[numerical_cols].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.4f')
plt.title('Correlation Matrix of Nifty Features')
plt.show()
```

```python

corr1 = df_new['close'].corr(df_new['open_interest'])
print(f"Correlation between Close Price and Open Interest: {corr1:.4f}")


corr2 = df_new['volume'].corr(df_new['open_interest'])
print(f"Correlation between Volume and Open Interest: {corr2:.4f}")


fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sns.scatterplot(ax=axes[0], data=df_new, x='open_interest', y='close', alpha=0.5, color='blue')
axes[0].set_title(f'Close Price vs Open Interest (r = {corr1:.4f})')

sns.scatterplot(ax=axes[1], data=df_new, x='open_interest', y='volume', alpha=0.5, color='green')
axes[1].set_title(f'Volume vs Open Interest (r = {corr2:.4f})')

plt.tight_layout()
plt.show()
```
