```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats
```

```python
df = pd.read_csv('/content/NIFTY-01FEB24-01JAN24.csv')
display(df.head())
```

```python
print("--- Summary Statistics of Nifty Dataset ---")
display(df.describe())
```

```python
print("--- Dataset Information ---")
df.info()
```

```python
print("--- Visualizing Distribution of Closing Price by Options Right (CE/PE) ---")
plt.figure(figsize=(10, 6))
sns.boxplot(x='right', y='close', data=df)
plt.title('Closing Price Distribution by Option Type (CE vs PE)')
plt.show()
```

```python
print("--- Strike Price Distribution ---")
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='strike_price', hue='right', kde=True, bins=30, multiple='stack')
plt.title('Strike Price Distribution for CE and PE Options')
plt.show()
```

```python

display(df.isnull().sum())
```

```python

print("Average Open Interest:")
display(df.groupby('right')['open_interest'].mean())
```

```python

print("Correlation Matrix:")
display(df[['open', 'high', 'low', 'close', 'open_interest', 'volume']].corr())
```

```python

ce_close = df[df['right'] == 'CE']['close']
pe_close = df[df['right'] == 'PE']['close']
print("Average Closing Price for CE:", ce_close.mean())
print("Average Closing Price for PE:", pe_close.mean())
```

```python

plt.figure(figsize=(12, 6))
sns.countplot(x='strike_price', hue='right', data=df[df['strike_price'].isin(df['strike_price'].value_counts().index[:10])])
plt.title('Top 10 Strike Prices Count by Option Type (CE/PE)')
plt.xticks(rotation=45)
plt.show()
```
