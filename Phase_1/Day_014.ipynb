```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


df_new = pd.read_csv('/content/NIFTY-01FEB24-24JAN24.csv')

display(df_new.head())
display(df_new.info())
```

```python

numerical_cols = ['open', 'high', 'low', 'close', 'open_interest', 'volume', 'strike_price']
corr_matrix = df_new[numerical_cols].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.4f')
plt.title('Correlation Matrix of Nifty Features')
plt.show()
```

```python

fig, axes = plt.subplots(2, 2, figsize=(16, 12))


corr_close = df_new['open_interest'].corr(df_new['close'])
sns.scatterplot(ax=axes[0, 0], data=df_new, x='open_interest', y='close', alpha=0.5, color='blue')
axes[0, 0].set_title(f'Close Price vs Open Interest (r = {corr_close:.4f})')


corr_volume = df_new['open_interest'].corr(df_new['volume'])
sns.scatterplot(ax=axes[0, 1], data=df_new, x='open_interest', y='volume', alpha=0.5, color='green')
axes[0, 1].set_title(f'Volume vs Open Interest (r = {corr_volume:.4f})')


corr_strike = df_new['open_interest'].corr(df_new['strike_price'])
sns.scatterplot(ax=axes[1, 0], data=df_new, x='open_interest', y='strike_price', alpha=0.5, color='purple')
axes[1, 0].set_title(f'Strike Price vs Open Interest (r = {corr_strike:.4f})')


corr_high = df_new['open_interest'].corr(df_new['high'])
sns.scatterplot(ax=axes[1, 1], data=df_new, x='open_interest', y='high', alpha=0.5, color='orange')
axes[1, 1].set_title(f'High Price vs Open Interest (r = {corr_high:.4f})')

plt.tight_layout()
plt.show()
```
