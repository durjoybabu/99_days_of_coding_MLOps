```python

import matplotlib.pyplot as plt
import seaborn as sns


sns.set_theme(style="whitegrid")


plt.figure(figsize=(10, 5))
sns.histplot(df['close'], bins=50, kde=True, color='blue')
plt.title('Distribution of NIFTY Close Prices')
plt.xlabel('Close Price')
plt.ylabel('Frequency')
plt.show()

plt.figure(figsize=(10, 5))
sns.scatterplot(data=df.sample(min(1000, len(df))), x='strike_price', y='close', hue='right', alpha=0.6)
plt.title('Strike Price vs Close Price (Sampled)')
plt.show()
```
